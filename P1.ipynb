## Phase 2: Data Loading

## Phase 3: Data Analysis

```
%%sql -r task11_result
-- Task 11: Display all customer details
SELECT * FROM CUSTOMER_SALES_DB.SALES_SCHEMA.CUSTOMERS;
```

```
%%sql -r task12_result
-- Task 12: Display all food item details
SELECT * FROM CUSTOMER_SALES_DB.SALES_SCHEMA.FOODITEMS;
```

```
%%sql -r task13_result
-- Task 13: Display all order details
SELECT * FROM CUSTOMER_SALES_DB.SALES_SCHEMA.ORDERS;
```

```
%%sql -r task14_result
-- Task 14: Customer-wise Sales Report
SELECT
    c.customer_id,
    c.first_name || ' ' || c.last_name AS customer_name,
    SUM(o.total_amount) AS total_spent
FROM CUSTOMER_SALES_DB.SALES_SCHEMA.CUSTOMERS c
INNER JOIN CUSTOMER_SALES_DB.SALES_SCHEMA.ORDERS o ON c.customer_id = o.customer_id
GROUP BY c.customer_id, customer_name
ORDER BY c.customer_id;
```

```
%%sql -r task15_result
-- Task 15: Highest Spending Customer
SELECT
    c.customer_id,
    c.first_name || ' ' || c.last_name AS customer_name,
    SUM(o.total_amount) AS total_spent
FROM CUSTOMER_SALES_DB.SALES_SCHEMA.CUSTOMERS c
INNER JOIN CUSTOMER_SALES_DB.SALES_SCHEMA.ORDERS o ON c.customer_id = o.customer_id
GROUP BY c.customer_id, customer_name
ORDER BY total_spent DESC
LIMIT 1;
```

```
%%sql -r task16_result
-- Task 16: Total Business Revenue
SELECT SUM(total_amount) AS total_revenue FROM CUSTOMER_SALES_DB.SALES_SCHEMA.ORDERS;
```

```
%%sql -r task17_result
-- Task 17: Category-wise Revenue Report
SELECT
    f.category,
    SUM(o.total_amount) AS total_revenue
FROM CUSTOMER_SALES_DB.SALES_SCHEMA.FOODITEMS f
INNER JOIN CUSTOMER_SALES_DB.SALES_SCHEMA.ORDERS o ON f.food_id = o.food_id
GROUP BY f.category
ORDER BY total_revenue DESC;
```

```
%%sql -r task18_result
-- Task 18: Order Status-wise Revenue Report
SELECT
    status AS order_status,
    SUM(total_amount) AS total_revenue
FROM CUSTOMER_SALES_DB.SALES_SCHEMA.ORDERS
GROUP BY status
ORDER BY total_revenue DESC;
```

```
%%sql -r task19_result
-- Task 19: Top Three Customers by total spending
SELECT
    ROW_NUMBER() OVER (ORDER BY SUM(o.total_amount) DESC) AS rank,
    c.first_name || ' ' || c.last_name AS customer_name,
    SUM(o.total_amount) AS total_spent
FROM CUSTOMER_SALES_DB.SALES_SCHEMA.CUSTOMERS c
INNER JOIN CUSTOMER_SALES_DB.SALES_SCHEMA.ORDERS o ON c.customer_id = o.customer_id
GROUP BY customer_name
ORDER BY total_spent DESC
LIMIT 3;
```

```
%%sql -r task20_result
-- Task 20: Customer Purchase Frequency Report
SELECT
    c.customer_id,
    c.first_name || ' ' || c.last_name AS customer_name,
    COUNT(o.order_id) AS orders_placed
FROM CUSTOMER_SALES_DB.SALES_SCHEMA.CUSTOMERS c
INNER JOIN CUSTOMER_SALES_DB.SALES_SCHEMA.ORDERS o ON c.customer_id = o.customer_id
GROUP BY c.customer_id, customer_name
ORDER BY orders_placed DESC, c.customer_id;
```

```
%%sql -r task21_result
-- Task 21: Display all Delivered Orders only
SELECT
    order_id, customer_id, food_id, status, total_amount
FROM CUSTOMER_SALES_DB.SALES_SCHEMA.ORDERS
WHERE status = 'Delivered';
```

```
%%sql -r task22_result
-- Task 22: Orders placed after 12 July 2026
SELECT
    o.order_id,
    c.first_name || ' ' || c.last_name AS customer_name,
    o.order_date,
    o.status,
    o.total_amount
FROM CUSTOMER_SALES_DB.SALES_SCHEMA.ORDERS o
INNER JOIN CUSTOMER_SALES_DB.SALES_SCHEMA.CUSTOMERS c ON o.customer_id = c.customer_id
WHERE o.order_date > '2026-07-12'
ORDER BY o.order_id;
```

## Phase 4: Views

```
%%sql -r task23_result
-- Task 23: Create a View named CUSTOMER_SALES_REPORT
CREATE OR REPLACE VIEW CUSTOMER_SALES_DB.SALES_SCHEMA.CUSTOMER_SALES_REPORT AS
SELECT
    c.customer_id,
    c.first_name || ' ' || c.last_name AS customer_name,
    SUM(o.total_amount) AS total_spent
FROM CUSTOMER_SALES_DB.SALES_SCHEMA.CUSTOMERS c
INNER JOIN CUSTOMER_SALES_DB.SALES_SCHEMA.ORDERS o ON c.customer_id = o.customer_id
GROUP BY c.customer_id, customer_name;
```

```
%%sql -r task24_result
-- Task 24: Retrieve all records from the View
SELECT * FROM CUSTOMER_SALES_DB.SALES_SCHEMA.CUSTOMER_SALES_REPORT;
```

```
%%sql -r task25_result
-- Task 25: Sort View data in descending order of Total Amount Spent
SELECT * FROM CUSTOMER_SALES_DB.SALES_SCHEMA.CUSTOMER_SALES_REPORT
ORDER BY total_spent DESC;
```
